```python
#name variables with full naming scheme 
# copy from modeling script + combine 
input_file = 'sulfonate_properties_pca' #a file with smiles, descriptors and IDs 
sheet_name = 'monosulfonates_dft_worked_avgO'
feat_start = 8 #column where descriptors start, 0-indexed
header_rows = 0 # adjust as needed 

id_colname= 'name' 
smiles_colname= 'SMILES' 
type_colname = 'Worked' #if you don't need to plot any subsets enter: None (not 'None')

#_______________________________________________________________________________
#_______________________________________________________________________________
inp = pd.read_excel(input_file+'.xlsx', sheet_name=sheet_name, header=header_rows)
inp_shape = inp.shape
allowed_empty = [smiles_colname] 
inp.dropna(subset=[col for col in inp.columns if col not in allowed_empty], inplace=True) #drop columns with empty values, allow empty smiles
inp_nona = inp.shape
if inp_shape[0] != inp_nona[0]:
    print('Some rows with missing values were removed')
    print('Original number of datapoints: ', inp_shape[0])
    print('New number of datapoints: ', inp_nona[0])

feats = inp.iloc[:,feat_start:] #just descriptors 
data = cf.get_mol_svg(inp) #sets up for interactive plots

ID = data[id_colname]
smiles = data[smiles_colname]
image = data['mol_image']

if type_colname is not None:
   typecol = data[type_colname]
if type_colname is None:
   typecol = None
```

```python
# take this with a grain of salt, a mathematically optimal UMAP is not necessarily the best UMAP for each case
# this is a good indicator of how well the UMAP is preserving the local structure of the data
# adjust functions to allow viewing of each UMAP hyperparameter combo so the user can choose 
# link to website explaining hyperparameters 

#let user define grid search parameters
def evaluate_clustering(features, umap_params, hdbscan_params):
    reducer = umap.UMAP(**umap_params)
    embedding = reducer.fit_transform(features)


    # Evaluate clustering with DB score
    if len(set(cluster_labels)) > 1:  # More than one cluster
        db_score = davies_bouldin_score(embedding, cluster_labels)
    else:
        db_score = np.inf
    
    return db_score, cluster_labels, embedding

def grid_search(features, umap_param_grid, hdbscan_param_grid):
    best_db_score = np.inf
    best_params = None
    best_labels = None
    best_embedding = None

    for umap_params in umap_param_grid:
        for hdbscan_params in hdbscan_param_grid:
            db_score, cluster_labels, embedding = evaluate_clustering(features, umap_params, hdbscan_params)
            
            # Optimization criteria
            if db_score < best_db_score:
                best_db_score = db_score
                best_params = (umap_params, hdbscan_params)
                best_labels = cluster_labels
                best_embedding = embedding

    return best_params, best_labels, best_embedding, best_db_score

umap_param_grid = [
    {'n_neighbors': n, 'min_dist': d, 'metric': 'euclidean'}
    for n in [5, 10, 15, 30, 50]
    for d in [0.0, 0.1, 0.3]
]

best_params, best_labels, best_embedding, best_db_score = grid_search(scaled_feats, umap_param_grid)

print(f'Best UMAP params: {best_params[0]}')
print(f'Best Davies-Bouldin score: {best_db_score}')

# Create a DataFrame with the UMAP results, original IDs, and cluster labels
umap_df = pd.DataFrame(best_embedding, columns=['UMAP1', 'UMAP2'])
umap_df['ID'] = ID
umap_df['Type'] = typecol
umap_df['Cluster'] = best_labels

# Plotting UMAP with HDBSCAN clusters, setting noise points to grey
plt.figure(figsize=(10, 8))

# Custom palette to include grey for noise
palette = sns.color_palette('viridis', as_cmap=True)
cluster_palette = {label: palette(i) for i, label in enumerate(set(best_labels) - {-1})}
cluster_palette[-1] = (0.5, 0.5, 0.5)  # Grey color for noise

sns.scatterplot(x='UMAP1', y='UMAP2', hue='Cluster', palette=cluster_palette, data=umap_df, legend='full')
plt.title(f'Optimized UMAP projection with HDBSCAN clusters')
plt.show()
```

```python
# see silhouette scores for different numbers of clusters
# ideal plot: clusters have above average silhouette scores, size of the silhouette plots are similar 
# and thickness of the plot for each cluster is uniform
# higher score (up to 1) indicates better clustering

n_clusters_range = range(2, 10) 
for k in n_clusters_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    cluster_labels = kmeans.fit_predict(clustering_data)
    
    silhouette_avg = silhouette_score(clustering_data, kmeans.fit_predict(clustering_data))
    
    fig, ax = plt.subplots(figsize=(8, 6))
    visualizer = SilhouetteVisualizer(kmeans, colors='yellowbrick', ax=ax)
    visualizer.fit(clustering_data)
    
    plt.xlabel('Silhouette Score')
    plt.title(f'Silhouette Score for {k} Clusters: {silhouette_avg:.2f}')
    plt.show()
    
```

```python
def scatter_plot(
    dataframe: pd.DataFrame, 
    x: str,  # x-axis data
    y: str,  # y-axis data
    title: str = 'default',  # title of plot
    x_label: str = 'default',  # axis label to be printed on plot (does not need to match dataframe name)
    x_range: tuple = None,  # range of x-axis
    y_label: str = 'default',  # axis label to be printed on plot (does not need to match dataframe name)
    y_range: tuple = None,  # range of y-axis
    legend: str = '',  # string with data label if using classifiers/building plots by category
    svgs: str = None,  # string with column name of svgs 
    hover_list: list = None,  # list of column names with data to be shown on hover 
    marker: str = 'o',  # marker type - most of the matplotlib markers are supported (https://matplotlib.org/stable/api/markers_api.html)
    color: str = '#931319',  # color of markers
    line_color: str = None,  # color of marker line_color
    alpha: int = 1,  # transparency of markers
    groupby: str = None,  # string with column name to group data by
    height: int = 500,  #plot height (recommended: 500)
    width: int = 500,  #plot width (recommended: 500)
    size: int = 10,  # size of markers (recommended: 10-20)
):

    if x_label == 'default':  # if no x_label provided, use x column name
        x_label = x
    if y_label == 'default':  # if no y_label provided, use y column name
        y_label = y

    if not x_range:
        x_min = min(dataframe[x]); x_max = max(dataframe[x])
        x_buffer = abs(x_max-x_min)/10
        x_range = (x_min-x_buffer, x_max+x_buffer)
    if not y_range:
        y_min = min(dataframe[y]); y_max = max(dataframe[y])
        y_buffer = abs(y_max-y_min)/10
        y_range = (y_min-y_buffer, y_max+y_buffer)

    if groupby is not None:
        hover_list = hover_list or []
        hover_list.insert(0, groupby)

    if svgs == None and labels == None: # no hover information provided
        if title == 'default':  # if no title provided, define from x, y labels
            title = f'{y_label} vs. {x_label}'
        plt = hv.Scatter(dataframe, kdims=[x], vdims=[y], label=legend).opts(title=title, marker=marker, height=height, width=width, color=color, alpha=alpha, size=size, line_color=line_color)
        
    else:  # hover information provided, build list of hover tools
        hover_list.insert(0, y)
        tooltips = f'<div>end' # beginning of tooltips if no svgs provided
        if svgs != None:
            tooltips = f'<div><div>@{svgs}{{safe}}</div>end'  # beginning of tooltips if svgs are provided
            hover_list.insert(1, svgs)
        if len(hover_list) < 4:
            for label in hover_list:
                if label != svgs and label != y:
                    tooltips = tooltips.replace('end', f'<div><span style="font-size: 17px; font-weight: bold;">@{label}</span></div>end')
        else:
            for label in hover_list:
                if label != svgs and label != y:
                    tooltips = tooltips.replace('end', f'<div><span style="font-size: 12px;">{label}: @{label}</span></div>end')
        
        tooltips = tooltips.replace('end', '</div>')
        hover = HoverTool(tooltips=tooltips)
        
        if title == 'default':  # if no title provided, define from x, y labels
            title = f'{y_label} vs. {x_label}'          
        plt = hv.Scatter(dataframe, kdims=[x], vdims=hover_list, label=legend).opts(title=title, marker=marker, height=height, width=width, tools=[hover], color=color, alpha=alpha, size=size, line_color=line_color)
        
        if groupby != None:
            # color = hv.Cycle(color).values
            plt = plt.opts(color=groupby, cmap=color)

        return plt
```

```python
n_clusters_range = range(2, 20) 
inertia_values = []
silhouette_scores = []

for n_clusters in n_clusters_range:
    birch_model = Birch(n_clusters=n_clusters, threshold=.5)
    birch_model.fit(clustering_data)
    labels = birch_model.predict(clustering_data)
    
    # Calculate the inertia (sum of squared distances to the closest centroid)
    # Note: BIRCH doesn't have an explicit inertia attribute, but we can calculate it manually
    cluster_centers = birch_model.subcluster_centers_
    inertia = np.sum([np.min(np.sum((point - cluster_centers)**2, axis=1)) for point in clustering_data])
    inertia_values.append(inertia)
    
    # Calculate the silhouette score
    silhouette_avg = silhouette_score(clustering_data, labels)
    silhouette_scores.append(silhouette_avg)
    

# Silhouette Score Plot
plt.subplot(1, 2, 2)
plt.plot(n_clusters_range, silhouette_scores, 'bo-')
plt.xlabel('Number of Clusters (n_clusters)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score for Optimal n_clusters')

plt.show()

```

## HCS Algorithm Clustering

```python
similarities = -pairwise_distances(clustering_data, metric='euclidean')
threshold = np.percentile(similarities, 75)

G = nx.Graph()
for i in range(clustering_data.shape[0]):
    G.add_node(i)
for i in range(clustering_data.shape[0]):
    for j in range(i + 1, clustering_data.shape[0]):
        if similarities[i, j] >= threshold:
            G.add_edge(i, j, weight=similarities[i, j])
```

```python
def is_highly_connected(subgraph):
    """
    Determine if a subgraph is highly connected.
    A graph is considered highly connected if the minimum cut of the graph is 
    greater than half the number of nodes in the graph.
    """
    n = len(subgraph.nodes)
    if n <= 2:
        return True
    cut_value, partition = nx.stoer_wagner(subgraph)
    return cut_value > n / 2

def hcs(graph):
    """
    HCS algorithm to find clusters within the graph.
    """
    if is_highly_connected(graph):
        return [graph]
    
    # Find the minimum cut
    cut_value, (partition_1, partition_2) = nx.stoer_wagner(graph)
    
    # Create subgraphs from the partitions
    subgraph_1 = graph.subgraph(partition_1)
    subgraph_2 = graph.subgraph(partition_2)
    
    # Recursively apply HCS to subgraphs
    clusters = []
    clusters.extend(hcs(subgraph_1))
    clusters.extend(hcs(subgraph_2))
    
    return clusters
```

```python
clusters = hcs(G)
labels = np.zeros(clustering_data.shape[0], dtype=int)
for cluster_id, cluster in enumerate(clusters):
    for node in cluster.nodes:
        labels[node] = cluster_id

# Add cluster labels to your data
df['Cluster'] = labels

# Visualize the clustering results
plt.figure(figsize=(10, 6))
sns.scatterplot(x='TSNE1', y='TSNE2', hue='Cluster', data=df, palette='viridis', s=100, edgecolor='k')
plt.title('HCS Clustering on 2D t-SNE Data')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.legend(title='Cluster')
plt.show()
```
